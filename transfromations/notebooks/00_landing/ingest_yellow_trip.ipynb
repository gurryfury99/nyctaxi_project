```python
import requests
import os

months = [
    "2025-12",
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04"
]

base_url = "https://d37ci6vzurychx.cloudfront.net/trip-data"
base_path = "/Volumes/nyctaxi_workspace/00_landing/data_sources/nyctaxi_yellow"

for month in months:

    file_name = f"yellow_tripdata_{month}.parquet"
    url = f"{base_url}/{file_name}"

    destination_dir = f"{base_path}/{month}"
    destination = f"{destination_dir}/{file_name}"

    os.makedirs(destination_dir, exist_ok=True)

    response = requests.get(url)

    if response.status_code == 200:
        with open(destination, "wb") as f:
            f.write(response.content)

        print(f"{month} - downloaded successfully")

    else:
        print(f"{month} - download failed: {response.status_code}")
```
