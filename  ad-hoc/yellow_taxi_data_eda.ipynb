how do daily trip volume and revenue trend over time?

```python
%sql
SELECT
    DATE(tpep_pickup_datetime) AS trip_date,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    ROUND(AVG(total_amount), 2) AS avg_revenue_per_trip,
    ROUND(AVG(trip_distance), 2) AS avg_trip_distance
FROM nyctaxi_workspace.nyctaxi_02_silver.yellow_trips_enriched
GROUP BY DATE(tpep_pickup_datetime)
ORDER BY trip_date;
```

which pickup borough contributes the most revenue,and how does demand compare?

```python
%sql
SELECT
    pu_borough,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    ROUND(AVG(total_amount), 2) AS avg_trip_revenue,
    ROUND(AVG(trip_distance), 2) AS avg_trip_distance
FROM nyctaxi_workspace.nyctaxi_02_silver.yellow_trips_enriched
GROUP BY pu_borough
ORDER BY total_revenue DESC;
```

how does revenue vary across hours of the day?

```python
%sql
SELECT
    HOUR(tpep_pickup_datetime) AS pickup_hour,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    ROUND(AVG(total_amount), 2) AS avg_revenue_per_trip,
    ROUND(AVG(trip_distance), 2) AS avg_trip_distance
FROM nyctaxi_workspace.nyctaxi_02_silver.yellow_trips_enriched
GROUP BY HOUR(tpep_pickup_datetime)
ORDER BY pickup_hour;
```

which pickup-to-dropoff borough routes are the most valuable?

```python
%sql
SELECT
    pu_borough,
    do_borough,
    COUNT(*) AS total_trips,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    ROUND(AVG(total_amount), 2) AS avg_trip_revenue,
    ROUND(AVG(trip_distance), 2) AS avg_trip_distance
FROM nyctaxi_workspace.nyctaxi_02_silver.yellow_trips_enriched
GROUP BY pu_borough, do_borough
HAVING COUNT(*) > 100
ORDER BY total_revenue DESC
LIMIT 20;
```
